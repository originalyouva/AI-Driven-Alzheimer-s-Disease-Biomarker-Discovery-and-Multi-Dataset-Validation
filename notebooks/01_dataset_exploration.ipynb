```python
sample_names = expression_data.columns[1:]

phenotype = []

for sample in sample_names:
    phenotype.append({
        "Sample": sample
    })

import pandas as pd

phenotype = pd.DataFrame(phenotype)

phenotype.head()
```

```python
for line in lines:
    if line.startswith("!Sample_title"):
        titles = line.strip().split("\t")[1:]
        break

disease_status = []

for title in titles:

    title = title.replace('"', '')

    if "control" in title.lower():
        disease_status.append("Control")
    elif "affected" in title.lower():
        disease_status.append("Alzheimer")
    else:
        disease_status.append("Unknown")

phenotype["Title"] = [t.replace('"','') for t in titles]
phenotype["Disease"] = disease_status

phenotype.head()
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

X = expression_data.drop(columns=["ID_REF"])

X = X.T

print(X.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print(X_pca.shape)

print("Explained variance ratio:")
print(pca.explained_variance_ratio_)
```

```python
import matplotlib.pyplot as plt

pca_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "Disease": phenotype["Disease"]
})

plt.figure(figsize=(8,6))

for disease in pca_df["Disease"].unique():
    subset = pca_df[pca_df["Disease"] == disease]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=disease
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA of Alzheimer's Gene Expression Dataset")
plt.legend()

plt.show()
```

```python
regions = []

for title in phenotype["Title"]:

    if title.startswith("EC"):
        regions.append("EC")
    elif title.startswith("HIP"):
        regions.append("HIP")
    elif title.startswith("MTG"):
        regions.append("MTG")
    elif title.startswith("PC"):
        regions.append("PC")
    elif title.startswith("SFG"):
        regions.append("SFG")
    elif title.startswith("VCX"):
        regions.append("VCX")
    else:
        regions.append("Unknown")

phenotype["Region"] = regions

phenotype.head()
```

```python
ec_sample_ids = ec_samples["Sample"].tolist()

ec_expression = expression_data[
    ["ID_REF"] + ec_sample_ids
]

ec_expression.shape
```

```python
ec_expression.head()
```

```python
control_ids = ec_samples[
    ec_samples["Disease"] == "Control"
]["Sample"].tolist()

alz_ids = ec_samples[
    ec_samples["Disease"] == "Alzheimer"
]["Sample"].tolist()

print("Controls:", len(control_ids))
print("Alzheimer:", len(alz_ids))
```

```python
gene_stats = pd.DataFrame()

gene_stats["Gene"] = ec_expression["ID_REF"]

gene_stats["Control_Mean"] = ec_expression[
    control_ids
].mean(axis=1)

gene_stats["Alzheimer_Mean"] = ec_expression[
    alz_ids
].mean(axis=1)

gene_stats.head()
```

```python
import numpy as np

gene_stats["Log2FC"] = np.log2(
    (gene_stats["Alzheimer_Mean"] + 1) /
    (gene_stats["Control_Mean"] + 1)
)

gene_stats.head()
```

```python
gene_stats.sort_values(
    "Log2FC",
    ascending=False
).head(20)
```

```python
from scipy.stats import ttest_ind

p_values = []

for _, row in ec_expression.iterrows():

    control_values = row[control_ids]
    alz_values = row[alz_ids]

    _, p = ttest_ind(
        control_values,
        alz_values,
        equal_var=False
    )

    p_values.append(p)

gene_stats["p_value"] = p_values

gene_stats.head()
```

```python
ec_numeric = ec_expression_numeric.set_index("ID_REF")

gene_variance = ec_numeric.var(axis=1)

top_genes = gene_variance.sort_values(
    ascending=False
).head(2000).index

ec_top = ec_numeric.loc[top_genes]

ec_top.shape
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X_ec = ec_top.T

X_scaled = StandardScaler().fit_transform(X_ec)

pca = PCA(n_components=2)

X_pca_ec = pca.fit_transform(X_scaled)

print(pca.explained_variance_ratio_)
```

```python
import matplotlib.pyplot as plt

pca_df = pd.DataFrame({
    "PC1": X_pca_ec[:,0],
    "PC2": X_pca_ec[:,1],
    "Disease": ec_samples["Disease"].values
})

plt.figure(figsize=(8,6))

for disease in pca_df["Disease"].unique():

    subset = pca_df[pca_df["Disease"] == disease]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=disease
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("EC Samples PCA (Top 2000 Variable Genes)")
plt.legend()
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import LeaveOneOut, cross_val_score

X_ml = ec_top.T

y_ml = (
    ec_samples["Disease"]
    .map({"Control":0, "Alzheimer":1})
    .values
)

rf = RandomForestClassifier(
    n_estimators=500,
    random_state=42
)

loo = LeaveOneOut()

scores = cross_val_score(
    rf,
    X_ml,
    y_ml,
    cv=loo,
    scoring="accuracy"
)

print("Accuracy:", scores.mean())
```

```python
rf.fit(X_ml, y_ml)

feature_importance = pd.DataFrame({
    "Gene": ec_top.index,
    "Importance": rf.feature_importances_
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

feature_importance.head(20)
```
