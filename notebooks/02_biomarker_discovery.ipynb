```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create volcano dataframe
volcano = de_results.copy()

volcano["minus_log10_FDR"] = -np.log10(volcano["FDR"])

# Validated biomarkers
validated_probes = [
    "215116_s_at",   # DNM1
    "212473_s_at",   # MICAL2
    "213869_x_at",   # THY1
    "203431_s_at"    # ARHGAP32
]

probe_to_gene = {
    "215116_s_at": "DNM1",
    "212473_s_at": "MICAL2",
    "213869_x_at": "THY1",
    "203431_s_at": "ARHGAP32"
}

# Select biomarkers
highlight = volcano[
    volcano["Gene"].isin(validated_probes)
].copy()

# Plot
plt.figure(figsize=(12, 9))

plt.scatter(
    volcano["Log2FC"],
    volcano["minus_log10_FDR"],
    alpha=0.4,
    s=25
)

plt.scatter(
    highlight["Log2FC"],
    highlight["minus_log10_FDR"],
    s=180
)

# Add labels
for _, row in highlight.iterrows():

    gene_name = probe_to_gene[row["Gene"]]

    plt.annotate(
        gene_name,
        (row["Log2FC"], row["minus_log10_FDR"]),
        fontsize=11,
        xytext=(5,5),
        textcoords="offset points"
    )

# Significance threshold
plt.axhline(
    -np.log10(0.05),
    linestyle="--"
)

# Fold change thresholds
plt.axvline(-1, linestyle="--", alpha=0.6)
plt.axvline(1, linestyle="--", alpha=0.6)

plt.xlabel("Log2 Fold Change", fontsize=12)
plt.ylabel("-Log10(FDR)", fontsize=12)
plt.title(
    "Validated Alzheimer's Disease Biomarkers",
    fontsize=14
)

plt.tight_layout()

plt.savefig(
    "../figures/volcano_plot_final.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
